```python
#1
df_raw = data.copy()
data = df_raw.copy()
data
```

```python
#2
data = df_raw.copy()
data = data.rename(columns={'Description': 'ProductDescription'})
data.head()
```

```python
#3
data['Revenue'] = data['Quantity'] * data['UnitPrice']
data.head()
```

```python
#4
import numpy as np
data['TransactionType'] = np.select(
    [
        data['InvoiceNo'].astype(str).str.startswith('C'),
        data['Quantity'] < 0,
        data['Quantity'] > 0
    ],
    [
        'Cancellation',
        'Return',
        'Positive Sale'
    ],
    default='Other/Invalid'
)
data['TransactionType'].value_counts()
```

```python
#5
data['IsIdentifiedCustomer'] = data['CustomerID'].notna()
data.head()
```

```python
data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'])
data.info()
```

```python
data['InvoiceNo'] = data['InvoiceNo'].astype('string')
data['StockCode'] = data['StockCode'].astype('string')
data['Country'] = data['Country'].astype('string')
data.dtypes
```

```python
cols = [
    'InvoiceNo', 'InvoiceDate', 'StockCode', 'ProductDescription',
    'CustomerID', 'Country', 'Quantity', 'UnitPrice', 'Revenue',
    'TransactionType', 'IsIdentifiedCustomer'
]
data = data[[col for col in cols if col in data.columns]]
data.head()
```

```python
data = data.sort_values(by=['InvoiceDate', 'InvoiceNo'])
data.head()
```

```python
data = data.reset_index(drop=True)
print(data.index)
print(data.index.is_unique)
print(data.index.equals(pd.RangeIndex(start=0, stop=len(data))))
```

```python
#1
missing_count = data.isnull().sum()
missing_percentage = (missing_count / len(data)) * 100

missing_report = pd.DataFrame({
    'MissingCount': missing_count,
    'MissingPercentage': missing_percentage
})
missing_report
```

```python
missing_report.sort_values(
    by='MissingPercentage',
    ascending=False
)
```

```python
data[data['CustomerID'].isna()]
```

```python
data[data['ProductDescription'].isna() | data['ProductDescription'].str.strip().eq('')]
```

```python
missing_report = pd.DataFrame({
    'ColumnName': data.columns,
    'MissingCount': data.isnull().sum().values,
    'NonMissingCount': data.notnull().sum().values,
    'MissingPercentage': (data.isnull().mean() * 100).values
})
missing_report
```

```python
data.duplicated().sum()
```

```python
data_clean = data.drop_duplicates().copy()
print("Records before cleaning:", len(data))
print("Records after cleaning:", len(data_clean))
```

```python
data.groupby(['InvoiceNo', 'StockCode']).size().reset_index(name='Count').query('Count > 1')
```

```python
data_clean['ProductDescription'] = data_clean['ProductDescription'].replace(r'^\s*$', pd.NA, regex=True)
data_clean['ProductDescription'] = data_clean['ProductDescription'].fillna('Unknown Description')
data_clean['CustomerID'].isna().sum()
```

```python
data_quality_report = pd.DataFrame({
    'Metric': [
        'Row Count',
        'Missing Values',
        'Exact Duplicate Rows'
    ],
    'Before Cleaning': [
        len(data),
        data.isna().sum().sum(),
        data.duplicated().sum()
    ],
    'After Cleaning': [
        len(data_clean),
        data_clean.isna().sum().sum(),
        data_clean.duplicated().sum()
    ]
})
data_quality_report
```

```python
data_quality_report.to_csv("data_quality_report.csv",index=False)
print("Data quality report saved succcessfully!")
```
